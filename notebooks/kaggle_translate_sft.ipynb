# Regional-language SFT data with IndicTrans2 (Kaggle GPU)

Translates a slice of the English training examples into Marathi, Bengali, Telugu and Tamil (change `LANGS`) with `ai4bharat/indictrans2-en-indic-1B`, keeping tool calls and numbers in English, and drops sentences whose round-trip chrF is poor.

Attach the `train.jsonl` dataset; output goes to `/kaggle/working/translated.jsonl` (download it into `data/llm/`, then re-run `python -m ml.llm.dedup_filter --keep-eval ...` with it as an extra input).

```python
REPO = "https://github.com/shivpratapsinghpanwar/KrishiDisha.ai.git"
BRANCH = "main"
LANGS = ["mr", "bn", "te", "ta"]
PER_LANG = 2500

import os, subprocess, sys, glob
if not os.path.exists("/kaggle/working/krishidisha"):
    subprocess.run(["git", "clone", "--depth", "1", "-b", BRANCH, REPO, "/kaggle/working/krishidisha"], check=True)
os.chdir("/kaggle/working/krishidisha")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "git+https://github.com/VarunGumma/IndicTransToolkit.git", "sentencepiece", "sacrebleu", "pyyaml"], check=True)
train = glob.glob("/kaggle/input/**/train.jsonl", recursive=True)[0]
print(train)
```

```python
!python -m ml.llm.translate_sft --inp {train} --out /kaggle/working/translated.jsonl --langs {' '.join(LANGS)} --per-lang {PER_LANG}
```
